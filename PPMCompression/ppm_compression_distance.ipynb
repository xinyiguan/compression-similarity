## Comparison to IDyOM

```python
idyom_results = pd.read_table("idyom_test/output/0_0-cpitch-cpitch-nil-nil-melody-nil-nil-ltm-nil-t-nil-c-nil-t-t-x-similarity-mean-t-t.dat", sep=" ")
idyom_results[["c1.description", "c2.description", "similarity"]]
```

```python
with open("idyom_test/test_chorale_melodies.json", "r") as f:
    test_melodies = json.load(f)
```

```python
for id1, id2 in zip(idyom_results["c1.description"], idyom_results["c2.description"]):
    res = distance(test_melodies[id1], test_melodies[id2])
    print(res)
```
