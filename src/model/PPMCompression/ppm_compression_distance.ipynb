## Comparison to IDyOM

```python
idyom_results = pd.read_table("idyom_test/output/0_0-cpitch-cpitch-nil-nil-melody-nil-nil-ltm-nil-t-nil-c-nil-t-t-x-similarity-mean-t-t.dat", sep=" ")
idyom_results[["c1.description", "c2.description", "similarity"]]
```

```python
with open("idyom_test/test_chorale_melodies.json", "r") as f:
    test_melodies = json.load(f)
```

```python
for id1, id2 in zip(idyom_results["c1.description"], idyom_results["c2.description"]):
    res = distance(test_melodies[id1], test_melodies[id2])
    print(res)
```

```python
def d2(x, y, alphabet_size: int | None = None, aggregation: callable = np.mean) -> float:
    c_x = compress_seq_ltm(x, x, alphabet_size, aggregation)
    c_y = compress_seq_ltm(y, y, alphabet_size, aggregation)
    c_xy = compress_seq_ltm(x, y, alphabet_size, aggregation)
    c_yx = compress_seq_ltm(y, x, alphabet_size, aggregation)

    return max(c_xy, c_yx) / max(c_x, c_y)
```

```python
for id1, id2 in zip(idyom_results["c1.description"], idyom_results["c2.description"]):
    a = np.unique(test_melodies[id1] + test_melodies[id2]).size
    res = d2(test_melodies[id1], test_melodies[id2], a)
    print(res)
```
